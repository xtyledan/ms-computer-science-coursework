### Multi-level Data Exploration Overview

Multi-level data exploration refers to the process of examining and analyzing data at different levels of detail or abstraction to uncover patterns, insights, and relationships. It’s like zooming in and out of your data to see both the big picture and the finer details from high-level to low-level exploration.

#### High-Level Exploration – Look at broad summaries:
- Mean, median, min, max
- istograms, box plots
- Count of categories, missing data

#### Mid-Level Exploration – Compare groups or variables:
- Correlations between variables
- Group-by aggregations (e.g., average sales per region)
- Visual comparisons like bar plots or scatter plots

#### Low-Level Exploration – Dive deep into specific cases:
- Investigating outliers
- Looking at individual records
- Tracing data transformations

It’s often used in exploration data analysis (EDA), business intelligence dashboards, or machine learning prep, helping analysts or data scientists move between summaries and specifics to better understand the data.

### Visualization 1: Correlation Heatmap Between Countries (Top 10 by Total Deaths)

```python
def create_correlation_heatmap(data):
    from pathlib import Path
    from IPython.display import Image, display

    print("Aggregating total deaths by country...")

    # Create a copy and convert the date column
    working_data = data.copy()
    working_data["date"] = pd.to_datetime(
        working_data["date"],
        errors="coerce"
    )

    # Exclude regions and economic groups that are not countries
    aggregate_columns = {
        "World",
        "Africa",
        "Asia",
        "Europe",
        "European Union (27)",
        "North America",
        "South America",
        "Oceania",
        "High-income countries",
        "Upper-middle-income countries",
        "Lower-middle-income countries",
        "Low-income countries"
    }

    country_columns = [
        column for column in working_data.select_dtypes(
            include=[np.number]
        ).columns
        if column not in aggregate_columns
    ]

    # Use the original values to rank countries because the
    # preprocessed values were normalized
    original_data = pd.read_csv("data/COVID_total_deaths.csv")
    original_data["date"] = pd.to_datetime(
        original_data["date"],
        errors="coerce"
    )
    original_data = original_data.sort_values("date")

    valid_country_columns = [
        column for column in country_columns
        if column in original_data.columns
    ]

    # Total deaths are cumulative, so use the latest available values
    total_deaths_by_country = (
        original_data[valid_country_columns]
        .iloc[-1]
        .sort_values(ascending=False)
    )

    print("Selecting top 10 countries by total deaths...")

    top_countries = total_deaths_by_country.head(10).index.tolist()
    print(top_countries)

    print("Filtering the dataset for the top countries...")

    filtered_data = working_data[
        ["date"] + top_countries
    ].copy()

    print("Preparing country data for correlation...")

    country_data = (
        filtered_data
        .dropna(subset=["date"])
        .sort_values("date")
        .set_index("date")[top_countries]
    )

    print("Computing the correlation matrix...")

    correlation_matrix = country_data.corr()

    print("Correlation matrix shape:")
    print(correlation_matrix.shape)

    display(correlation_matrix.head())

    print("Plotting the heatmap...")

    output_folder = Path("output")
    output_folder.mkdir(exist_ok=True)

    heatmap_path = output_folder / "correlation_heatmap.png"

    plt.figure(figsize=(12, 10))

    sns.heatmap(
        correlation_matrix,
        annot=True,
        fmt=".2f",
        cmap="vlag",
        vmin=-1,
        vmax=1,
        center=0,
        square=True,
        linewidths=0.5,
        cbar_kws={"label": "Correlation"}
    )

    plt.title(
        "Correlation of Normalized COVID-19 Death Trends\n"
        "Among the Top 10 Countries",
        fontsize=15,
        fontweight="bold",
        pad=15
    )
    plt.xlabel("Country", fontsize=11)
    plt.ylabel("Country", fontsize=11)
    plt.xticks(rotation=45, ha="right")
    plt.yticks(rotation=0)
    plt.tight_layout()

    plt.savefig(
        heatmap_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    print(f"Heatmap saved to {heatmap_path}")
    display(Image(filename=str(heatmap_path)))


create_correlation_heatmap(preprocessed_data)
```

### Visualization 2: Hierarchical Bar Chart (Top 20 Countries by Total Deaths)

```python
def create_bar_chart(data):
    from pathlib import Path
    from IPython.display import Image, display
    from matplotlib.ticker import StrMethodFormatter

    print("Creating bar chart for top 20 countries...")

    # Use the original data because the preprocessed values are normalized
    chart_data = pd.read_csv("data/COVID_total_deaths.csv")

    chart_data["date"] = pd.to_datetime(
        chart_data["date"],
        errors="coerce"
    )

    chart_data = chart_data.sort_values("date")

    # Exclude regions and economic groups
    aggregate_columns = {
        "World",
        "Africa",
        "Asia",
        "Europe",
        "European Union (27)",
        "North America",
        "South America",
        "Oceania",
        "High-income countries",
        "Upper-middle-income countries",
        "Lower-middle-income countries",
        "Low-income countries"
    }

    country_columns = [
        column for column in chart_data.select_dtypes(
            include=[np.number]
        ).columns
        if column not in aggregate_columns
    ]

    # Total deaths are cumulative, so use the latest values
    total_deaths_by_country = pd.to_numeric(
        chart_data[country_columns].iloc[-1],
        errors="coerce"
    ).dropna()

    # Sort by total deaths and select the top 20 countries
    top_20_countries = (
        total_deaths_by_country
        .nlargest(20)
        .sort_values()
        .reset_index()
    )

    top_20_countries.columns = ["Country", "Total Deaths"]

    print("Top 20 countries selected:")
    display(top_20_countries.sort_values(
        "Total Deaths",
        ascending=False
    ))

    # Create the bar chart
    fig, ax = plt.subplots(figsize=(12, 10))

    colors = [
        "#1F4E79" if country == top_20_countries.iloc[-1]["Country"]
        else "#8FBBD9"
        for country in top_20_countries["Country"]
    ]

    bars = ax.barh(
        top_20_countries["Country"],
        top_20_countries["Total Deaths"],
        color=colors,
        edgecolor="white"
    )

    # Add value labels
    ax.bar_label(
        bars,
        labels=[
            f"{value:,.0f}"
            for value in top_20_countries["Total Deaths"]
        ],
        padding=5,
        fontsize=9
    )

    # Add labels and title
    ax.set_title(
        "Top 20 Countries by Total COVID-19 Deaths",
        fontsize=16,
        fontweight="bold",
        pad=15
    )
    ax.set_xlabel("Total Deaths", fontsize=11)
    ax.set_ylabel("Country", fontsize=11)
    ax.xaxis.set_major_formatter(
        StrMethodFormatter("{x:,.0f}")
    )
    ax.grid(
        axis="x",
        linestyle="--",
        alpha=0.3
    )
    ax.margins(x=0.18)

    plt.tight_layout()

    # Save the plot
    output_folder = Path("output")
    output_folder.mkdir(exist_ok=True)

    bar_chart_path = output_folder / "bar_chart.png"

    plt.savefig(
        bar_chart_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    print(f"Bar chart saved to {bar_chart_path}")

    # Display the saved image
    display(Image(filename=str(bar_chart_path)))


create_bar_chart(preprocessed_data)
```

### Visualization 3: Time Series Chart (Global Trends in Total Deaths)

```python
def create_time_series_chart(data):
    from pathlib import Path
    from IPython.display import Image, display
    from matplotlib.ticker import StrMethodFormatter
    import matplotlib.dates as mdates

    print("Creating global time series chart...")

    # Use the original data because the preprocessed values are normalized
    time_series_data = pd.read_csv(
        "data/COVID_total_deaths.csv"
    )

    time_series_data["date"] = pd.to_datetime(
        time_series_data["date"],
        errors="coerce"
    )

    time_series_data = (
        time_series_data
        .dropna(subset=["date"])
        .sort_values("date")
    )

    # Exclude regional and economic aggregates to prevent double counting
    aggregate_columns = {
        "World",
        "Africa",
        "Asia",
        "Europe",
        "European Union (27)",
        "North America",
        "South America",
        "Oceania",
        "High-income countries",
        "Upper-middle-income countries",
        "Lower-middle-income countries",
        "Low-income countries"
    }

    country_columns = [
        column for column in time_series_data.select_dtypes(
            include=[np.number]
        ).columns
        if column not in aggregate_columns
    ]

    # Aggregate total deaths across countries for each date
    time_series_data["Global Total Deaths"] = (
        time_series_data[country_columns]
        .fillna(0)
        .sum(axis=1)
    )

    global_deaths = time_series_data[
        ["date", "Global Total Deaths"]
    ].copy()

    # Create the time series chart
    fig, ax = plt.subplots(figsize=(14, 7))

    ax.plot(
        global_deaths["date"],
        global_deaths["Global Total Deaths"],
        color="#1F4E79",
        linewidth=2.5,
        label="Global Total Deaths"
    )

    ax.fill_between(
        global_deaths["date"],
        global_deaths["Global Total Deaths"],
        color="#8FBBD9",
        alpha=0.3
    )

    # Add milestone annotations
    milestones = [
        (1_000_000, "1 million deaths"),
        (5_000_000, "5 million deaths")
    ]

    for threshold, label in milestones:
        matching_rows = global_deaths[
            global_deaths["Global Total Deaths"] >= threshold
        ]

        if not matching_rows.empty:
            milestone = matching_rows.iloc[0]

            ax.annotate(
                label,
                xy=(
                    milestone["date"],
                    milestone["Global Total Deaths"]
                ),
                xytext=(15, 25),
                textcoords="offset points",
                fontsize=10,
                fontweight="bold",
                arrowprops={
                    "arrowstyle": "->",
                    "color": "#444444"
                }
            )

    # Annotate the latest total
    latest = global_deaths.iloc[-1]

    ax.annotate(
        f'{latest["Global Total Deaths"]:,.0f}',
        xy=(
            latest["date"],
            latest["Global Total Deaths"]
        ),
        xytext=(-10, 15),
        textcoords="offset points",
        ha="right",
        fontsize=10,
        fontweight="bold",
        color="#1F4E79"
    )

    # Add labels and title
    ax.set_title(
        "Global Cumulative COVID-19 Deaths Over Time",
        fontsize=16,
        fontweight="bold",
        pad=15
    )
    ax.set_xlabel("Date", fontsize=11)
    ax.set_ylabel("Total Deaths", fontsize=11)
    ax.legend(loc="upper left")

    ax.yaxis.set_major_formatter(
        StrMethodFormatter("{x:,.0f}")
    )
    ax.xaxis.set_major_locator(
        mdates.YearLocator()
    )
    ax.xaxis.set_major_formatter(
        mdates.DateFormatter("%Y")
    )

    ax.grid(
        linestyle="--",
        alpha=0.3
    )
    ax.margins(x=0.01, y=0.1)

    plt.tight_layout()

    # Save the plot
    output_folder = Path("output")
    output_folder.mkdir(exist_ok=True)

    time_series_path = output_folder / "time_series_chart.png"

    plt.savefig(
        time_series_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.close()

    print(f"Time series chart saved to {time_series_path}")

    # Display the saved image
    display(Image(filename=str(time_series_path)))


create_time_series_chart(preprocessed_data)
```

### Visualization 1: Drill-Down/Zoom Capabilities
<div class="alert alert-block alert-info">

```python
import pandas as pd
import numpy as np
import plotly.express as px

# Load the cleaned/preprocessed dataset
data = pd.read_csv("data/COVID_total_deaths_preprocessed.csv")


def create_drill_down_chart(data):
    print("\nCreating drill-down/zoom visualization...")

    # Make a copy of the dataset
    chart_data = data.copy()

    # Convert the date column to datetime
    chart_data["date"] = pd.to_datetime(
        chart_data["date"],
        errors="coerce"
    )

    # Remove invalid dates and sort chronologically
    chart_data = (
        chart_data
        .dropna(subset=["date"])
        .sort_values("date")
    )

    # Exclude regional and economic aggregates to prevent double counting
    aggregate_columns = {
        "World",
        "Africa",
        "Asia",
        "Europe",
        "European Union (27)",
        "North America",
        "South America",
        "Oceania",
        "High-income countries",
        "Upper-middle-income countries",
        "Lower-middle-income countries",
        "Low-income countries"
    }

    # Select only numeric country columns
    country_columns = [
        column
        for column in chart_data.select_dtypes(
            include=[np.number]
        ).columns
        if column not in aggregate_columns
    ]

    # Sum the normalized country values globally for each date
    global_deaths_by_date = chart_data[["date"]].copy()

    global_deaths_by_date["Global Death Measure"] = (
        chart_data[country_columns]
        .fillna(0)
        .sum(axis=1)
    )

    # Create an interactive time series chart with zoom capability
    figure = px.line(
        global_deaths_by_date,
        x="date",
        y="Global Death Measure",
        title="Interactive Global COVID-19 Death Trend",
        labels={
            "date": "Date",
            "Global Death Measure": "Combined Normalized Death Measure"
        }
    )

    # Add zoom buttons and range slider
    figure.update_xaxes(
        rangeslider_visible=True,
        rangeselector={
            "buttons": [
                {
                    "count": 1,
                    "label": "1 Month",
                    "step": "month",
                    "stepmode": "backward"
                },
                {
                    "count": 6,
                    "label": "6 Months",
                    "step": "month",
                    "stepmode": "backward"
                },
                {
                    "count": 1,
                    "label": "1 Year",
                    "step": "year",
                    "stepmode": "backward"
                },
                {
                    "step": "all",
                    "label": "All"
                }
            ]
        }
    )

    # Configure hover information
    figure.update_traces(
        line={"width": 3},
        hovertemplate=(
            "<b>Date:</b> %{x|%B %d, %Y}<br>"
            "<b>Death measure:</b> %{y:,.4f}"
            "<extra></extra>"
        )
    )

    # Configure chart layout
    figure.update_layout(
        template="plotly_white",
        title_x=0.5,
        hovermode="x unified",
        font={"size": 12},
        height=600
    )

    # Show the chart
    figure.show()


# Generate visualization
print("Generating visualizations...")

print("\nVisualization 1. Creating Drill-Down/Zoom Capabilities...")

# Call the create_drill_down_chart() function
create_drill_down_chart(data)
```

```python
import pandas as pd
import numpy as np
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Load the cleaned/preprocessed dataset
data = pd.read_csv("data/COVID_total_deaths_preprocessed.csv")


def create_overview_detail_chart(data):
    print("\nCreating overview+detail visualization...")

    # Make a copy of the dataset
    chart_data = data.copy()

    # Convert date column to datetime
    chart_data["date"] = pd.to_datetime(
        chart_data["date"],
        errors="coerce"
    )

    # Remove invalid dates and sort chronologically
    chart_data = (
        chart_data
        .dropna(subset=["date"])
        .sort_values("date")
    )

    # Exclude regional and economic aggregates
    aggregate_columns = {
        "World",
        "Africa",
        "Asia",
        "Europe",
        "European Union (27)",
        "North America",
        "South America",
        "Oceania",
        "High-income countries",
        "Upper-middle-income countries",
        "Lower-middle-income countries",
        "Low-income countries"
    }

    # Select numeric country columns
    country_columns = [
        column
        for column in chart_data.select_dtypes(
            include=[np.number]
        ).columns
        if column not in aggregate_columns
    ]

    # Aggregate data for the top 5 countries by total deaths
    country_totals = (
        chart_data[country_columns]
        .fillna(0)
        .sum()
        .sort_values(ascending=False)
    )

    top_5_countries = country_totals.head(5).index.tolist()

    print("Top 5 countries:")
    for country in top_5_countries:
        print(country)

    # Filter the dataset for the top 5 countries
    top_5_data = chart_data[
        ["date"] + top_5_countries
    ].copy()

    # Create an overview + detail chart
    figure = make_subplots(
        rows=2,
        cols=1,
        shared_xaxes=True,
        vertical_spacing=0.08,
        row_heights=[0.25, 0.75],
        subplot_titles=(
            "Overview",
            "Detailed View"
        )
    )

    # Overview portion of the chart
    for country in top_5_countries:
        figure.add_trace(
            go.Scatter(
                x=top_5_data["date"],
                y=top_5_data[country],
                mode="lines",
                name=country,
                legendgroup=country,
                showlegend=False,
                line={"width": 1}
            ),
            row=1,
            col=1
        )

    # Detail portion of the chart
    for country in top_5_countries:
        figure.add_trace(
            go.Scatter(
                x=top_5_data["date"],
                y=top_5_data[country],
                mode="lines",
                name=country,
                legendgroup=country,
                line={"width": 2},
                hovertemplate=(
                    "<b>%{fullData.name}</b><br>"
                    "Date: %{x|%B %d, %Y}<br>"
                    "Normalized deaths: %{y:.4f}"
                    "<extra></extra>"
                )
            ),
            row=2,
            col=1
        )

    # Add range slider to the overview chart
    figure.update_xaxes(
        rangeslider={"visible": True},
        row=1,
        col=1
    )

    # Configure layout
    figure.update_layout(
        title={
            "text": "COVID-19 Death Trends: Top 5 Countries",
            "x": 0.5
        },
        template="plotly_white",
        height=750,
        hovermode="x unified",
        legend_title_text="Country"
    )

    # Configure axis labels
    figure.update_yaxes(
        title_text="Normalized Deaths",
        row=1,
        col=1
    )

    figure.update_yaxes(
        title_text="Normalized Deaths",
        row=2,
        col=1
    )

    figure.update_xaxes(
        title_text="Date",
        row=2,
        col=1
    )

    # Show the chart
    figure.show()
```

## Load the Dataset

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display

# To Do: Load the original dataset `data/COVID_total_deaths.csv` from the 'data' folder.
data = pd.read_csv("data/COVID_total_deaths.csv")
```

## Clean and Preprocess the Data

```python
def clean_data(data):
    print("\nCleaning the dataset...")

    filtered_data = data.copy()

    # Convert 'date' column to datetime
    filtered_data["date"] = pd.to_datetime(
        filtered_data["date"],
        errors="coerce"
    )

    filtered_data = filtered_data.dropna(subset=["date"])

    country_columns = [
        column for column in filtered_data.columns
        if column != "date"
    ]

    # Convert country columns to numeric
    filtered_data[country_columns] = (
        filtered_data[country_columns]
        .apply(pd.to_numeric, errors="coerce")
        .fillna(0)
    )

    # To Do: Remove countries with all zero deaths
    non_zero_countries = [
        column for column in country_columns
        if filtered_data[column].ne(0).any()
    ]

    filtered_data = filtered_data[
        ["date"] + non_zero_countries
    ]

    # To Do: Remove rows (dates) with all zero deaths
    filtered_data = filtered_data[
        filtered_data[non_zero_countries]
        .ne(0)
        .any(axis=1)
    ]

    filtered_data = (
        filtered_data
        .sort_values("date")
        .reset_index(drop=True)
    )

    # Display filtered dataset shape
    print("Filtered dataset shape:", filtered_data.shape)

    return filtered_data
```

```python
# To Do: Call the clean_data() function.
cleaned_data = clean_data(data)
```

## Select Top 10 Countries

```python
def select_top_countries(data):
    print("\nSelecting top 10 countries by total deaths...")

    # Make a copy of the dataset
    filtered_data = data.copy()

    # To Do: Calculate total deaths by country.
    country_columns = [
        column for column in filtered_data.columns
        if column != "date"
    ]

    total_deaths = (
        filtered_data[country_columns]
        .fillna(0)
        .max()
    )

    # To Do: Exclude these aggregate regions or groups
    aggregate_regions = {
        "World",
        "Africa",
        "Europe",
        "North America",
        "South America",
        "Asia",
        "European Union (27)",
        "High-income countries",
        "Upper-middle-income countries",
        "Lower-middle-income countries"
    }

    total_deaths = total_deaths.drop(
        labels=[
            region for region in aggregate_regions
            if region in total_deaths.index
        ]
    )

    # To Do: Select top 10 individual countries
    top_10_countries = (
        total_deaths
        .sort_values(ascending=False)
        .head(10)
    )

    print("\nTop 10 individual countries by total deaths:")

    # To Do: Display the top 10 individual countries
    print(top_10_countries)

    # To Do: Filter the dataset for the top 10 countries
    filtered_data = filtered_data[
        ["date"] + top_10_countries.index.tolist()
    ].copy()

    # Ensure date column is datetime
    filtered_data["date"] = pd.to_datetime(
        filtered_data["date"],
        errors="coerce"
    )

    filtered_data = filtered_data.dropna(subset=["date"])

    # To Do: Aggregate data weekly (downsample)
    filtered_data = (
        filtered_data
        .set_index("date")
        .resample("W")
        .max()
        .reset_index()
    )

    # To Do: Display filtered dataset shape
    print(
        f"\nFiltered dataset shape after selecting top 10 countries: "
        f"{filtered_data.shape}"
    )

    # Return filtered data
    return filtered_data
```

```python
# To Do: Call select_top_countries()
top_countries_data = select_top_countries(cleaned_data)
```

## Prepare Data for Animation

```python
print("Preparing data for animation...")

# To Do: Melt the data for easier plotting
animation_data = top_countries_data.melt(
    id_vars=["date"],
    var_name="Country",
    value_name="Total Deaths"
)

animation_data = animation_data.sort_values(
    ["date", "Country"]
).reset_index(drop=True)
```

```python
print("Data prepared successfully!")

# To Do: Display melted data
display(animation_data.head(20))
```

## Create the Matplotlib Animation

```python
print("Creating the animation...ignore static output")

plt.style.use("seaborn-v0_8-whitegrid")

# To Do: Initialize the plot
fig, ax = plt.subplots(figsize=(12, 7))

countries = animation_data["Country"].unique()

dates = sorted(
    animation_data["date"].unique()
)

lines = {}

for country in countries:
    line, = ax.plot(
        [],
        [],
        linewidth=2,
        label=country
    )

    lines[country] = line

# To Do: Set the axis limits and labels
ax.set_xlim(
    animation_data["date"].min(),
    animation_data["date"].max()
)

ax.set_ylim(
    0,
    animation_data["Total Deaths"].max() * 1.10
)

ax.set_xlabel("Date")
ax.set_ylabel("Total Deaths")
ax.set_title("COVID-19 Death Trends for Top 10 Countries")

ax.grid(True, alpha=0.3)

ax.legend(
    title="Country",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

fig.tight_layout()
```

### Function to initialize the animation

```python
def init():
    # To Do: Create a for loop here to initialize lines values.
    for country in countries:
        lines[country].set_data([], [])

    # Return lines values
    return list(lines.values())
```

### Function to update the animation frame-by-frame

```python
def update(frame):
    # To Do: Update the animation frame-by-frame.

    current_date = dates[frame]

    current_data = animation_data[
        animation_data["date"] <= current_date
    ]

    for country in countries:
        country_data = current_data[
            current_data["Country"] == country
        ]

        lines[country].set_data(
            country_data["date"],
            country_data["Total Deaths"]
        )

    display_date = pd.Timestamp(
        current_date
    ).strftime("%B %d, %Y")

    ax.set_title(
        "COVID-19 Death Trends for Top 10 Countries\n"
        f"Date: {display_date}"
    )

    return list(lines.values())
```

### Create the animation

<div class="alert alert-block alert-info">
    <span style="color:black"><b><i>Hints:</i></b> Use <b><mark>FuncAnimation()</mark></b> to initate the animation visualization. Use <b><mark>HTML({variable}.to_jshtml())</mark></b> to render your animation inside the notebook in an html exportable format.</span>
</div>

```python
# To Do: Call FuncAnimation() to create the animation
# interval=200 means each frame will last 200 ms (0.2 sec)

ani = FuncAnimation(
    fig,
    update,
    frames=len(dates),
    init_func=init,
    interval=200,
    blit=False,
    repeat=False
)

# Prevent duplicate static output
plt.close(fig)

# To Do: Show the animation
HTML(ani.to_jshtml())
```
